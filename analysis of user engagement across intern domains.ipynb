```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("C:\\Users\\bhara\\Downloads\\internship_domain_engagement_yearly.csv")

# Count applications per domain
domain_counts = df['internship_domain'].value_counts()

# Plotting
plt.figure(figsize=(10, 6))
sns.barplot(x=domain_counts.index, y=domain_counts.values)
plt.title('Bar Graph representation of User Engagement by Internship Domain')
plt.xlabel('Internship Domain')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)

# Plot as pie chart
plt.figure(figsize=(8, 8))
plt.pie(domain_counts, labels=domain_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Pie chart representation of User Engagement by Internship Domain')
plt.axis('equal')  # Equal aspect ratio ensures pie is a circle
plt.show()



# Convert to datetime with day first
df['application_date'] = pd.to_datetime(df['application_date'], dayfirst=True)

# Group by year and domain
yearly_trend = df.groupby([df['application_date'].dt.year, 'internship_domain'])['student_id'].count().unstack().fillna(0)

# Plot
yearly_trend.plot(kind='bar', figsize=(12,6))
plt.title("Yearly Application Trends by Domain")
plt.xlabel("Year")
plt.ylabel("Number of Applications")
plt.xticks(rotation=0)
plt.legend(title="Domain", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

```
